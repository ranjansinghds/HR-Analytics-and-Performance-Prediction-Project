# TASK #3: VISUALIZE DATASET

```python
# Let's replace 'Attrition' , 'OverTime' and 'Over18' column with integers before performing any visualizations
employee_df['Attrition'] = employee_df['Attrition'].apply(lambda x: 1 if x == 'Yes' else 0)
employee_df['Over18'] = employee_df['Over18'].apply(lambda x: 1 if x == 'Y' else 0)
employee_df['OverTime'] = employee_df['OverTime'].apply(lambda x: 1 if x == 'Yes' else 0)
```

```python
# checking the dataset again after changes
employee_df[['Attrition', 'Over18', 'OverTime']].head()
```

```python
# Let's see if we have any missing values in the dataset
employee_df.isnull().sum()
```

```python
# we have checked about missing data and we don't have any missing data and we will aslo chweck it with visualization too
plt.figure(figsize=(10, 6))
sns.heatmap(employee_df.isnull(), cbar=False, cmap='viridis', yticklabels=False)
```

```python
# we are usning a histogram to check all the data of a dataset for more info visualy together
employee_df.hist(figsize=(20, 20), bins=30, edgecolor='black')

# Several feature such as 'MonthlyIncome' and 'TotalWorkingYears' are tail heavy
```

```python
# It makes sense to drop 'EmployeeCount', 'StandardHours' and 'Over18' columns since they do not change from one employee to the other
# Let's drop 'EmployeeNumber' as well
employee_df.drop(['EmployeeCount', 'StandardHours', 'Over18', 'EmployeeNumber'], axis=1, inplace=True) 
```
