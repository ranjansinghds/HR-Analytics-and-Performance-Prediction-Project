### Features that need special attention
**EmployeeCount:**
Every observation has a value of 1. It does not provide useful information for predicting attrition.

**Over18:**
Every employee belongs to the same category. This feature has no predictive variation.

**StandardHours:**
All employees have 80 standard hours. It does not help distinguish employees.

**EmployeeNumber:**
This is an employee identifier, not a meaningful measure of employee characteristics. It should generally be excluded from predictive modeling.

```python
# It makes sense to drop 'EmployeeCount', 'StandardHours' and 'Over18' columns since they do not change from one employee to the other
# Let's drop 'EmployeeNumber' as well
employee_df.drop(['EmployeeCount', 'StandardHours', 'Over18', 'EmployeeNumber'], axis=1, inplace=True) 
```

```python
# Job level is strongly correlated with total working hours
# Monthly income is strongly correlated with Job level
# Monthly income is strongly correlated with total working hours
# Age is stongly correlated with monthly income

correlations = employee_df.select_dtypes(include='number').corr()
plt.figure(figsize=(20, 20))
sns.heatmap(correlations, annot=True, cmap='viridis', fmt='.2f') # fmt='.2f' limits decimal places
```

**Analysis:**

The correlation heatmap helps identify relationships between numerical features and the target variable, Attrition.<br>

The analysis shows that OverTime has the strongest positive correlation with Attrition (0.25), suggesting that employees who work overtime may be more likely to leave the company.<br>

Several features, including Age, JobLevel, MonthlyIncome, TotalWorkingYears, and YearsInCurrentRole, have weak negative correlations with Attrition.<br>

Additionally, some independent features show strong positive correlations with each other. For example, JobLevel and MonthlyIncome have a correlation of 0.95, while Age and TotalWorkingYears have a correlation of 0.68.<br>

These relationships indicate potential multicollinearity among certain features, which should be considered during model development.

```python
# KDE (Kernel Density Estimate) is used for visualizing the probability Density of a continuous variable.
# KDE describes the probablity density at different values in a continuous variable.

# checking stayed and left employees with Distance From Home
plt.figure(figsize=(12,8))
sns.kdeplot(left_df['DistanceFromHome'], label='Employee who left', fill=True, color='red')
sns.kdeplot(stayed_df['DistanceFromHome'], label='Employee who stayed', fill=True, color='blue')

plt.xlabel('Distance from home')
plt.legend()
```

**Analysis**

The KDE plot shows differences in the distribution of commuting distances between employees who left and those who stayed. Employees who stayed have a higher concentration around shorter distances, particularly between 2 and 10 units. In contrast, employees who left show a relatively higher density across some longer commuting distances, especially between 12 and 26 units. This suggests that commuting distance may be associated with employee attrition. However, the considerable overlap between the two distributions indicates that distance from home alone may not explain employee departures.

```python
# checking stayed and left employees with Year with current manager

plt.figure(figsize=(12,8))
sns.kdeplot(left_df['YearsWithCurrManager'], label='Employee who left',  fill=True, color='red')
sns.kdeplot(stayed_df['YearsWithCurrManager'], label='Employee who stayed', fill=True, color='blue')

plt.xlabel('Years With Current Manager')
plt.legend()
```

**Analysis:**

The KDE plot shows that most employees who left the company had spent around 0–3 years working with their current manager. On the other hand, employees who stayed show a noticeable concentration around 2–3 years and another peak around 7 years. The distribution also indicates that employees who stayed are more represented among those who have worked with their current manager for longer periods. This suggests that the length of time employees work with their current manager may be associated with employee attrition.

```python
# checking stayed and left employees with Total working years

plt.figure(figsize=(12,7))
sns.kdeplot(left_df['TotalWorkingYears'], label='Employee who left', fill=True, color='red')
sns.kdeplot(stayed_df['TotalWorkingYears'], label='Employee who stayed', fill=True, color='blue')

plt.xlabel('Total Working Years')
plt.legend()
```

**Analysis:**

The KDE plot shows that employees who left the company generally had fewer years of total work experience, with the highest concentration around 5–7 years. In comparison, employees who stayed have a higher concentration around 8–9 years and are more represented among those with longer work experience. This suggests that employees with relatively less professional experience may be more likely to leave the company. However, both distributions overlap considerably, indicating that work experience alone does not explain employee attrition.

# TASK #4: CREATE TESTING AND TRAINING DATASET & PERFORM DATA CLEANING

```python
# Checking data again for cleaning 
employee_df.head(4)
```

```python
# Checking all categorical data
X_cat = employee_df.select_dtypes(include='object')
```

```python
X_cat
```

```python
# importing important library to change categorical data to number
from sklearn.preprocessing import OneHotEncoder

onehotencoder = OneHotEncoder()

X_cat = onehotencoder.fit_transform(X_cat).toarray()
```

```python
# checking the changes in data
X_cat.shape
```

```python
# converting this new data to a dataframe
X_cat = pd.DataFrame(X_cat)
```

```python
X_cat
```

```python
# note that we dropped the target 'Atrittion'
X_numerical = employee_df.select_dtypes(include='number').drop('Attrition', axis=1)
```

```python
# doing the Concatenation the all categorical and numerical dataframe together
X_all = pd.concat([X_cat, X_numerical], axis=1)
```

```python
X_all
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

# X_all is a DataFrame, convert its column names to strings before applying MinMaxScaker
X_all.columns = X_all.columns.astype(str)

X = scaler.fit_transform(X_all)
```

```python
X
```

```python
# now we are putting our target column in y variable from our employee dataset
y = employee_df['Attrition']
y
```

# TASK #5: TRAIN AND EVALUATE A LOGISTIC REGRESSION CLASSIFIER

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
# checking the tarin test split shape
X_train.shape
```

```python
X_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
# Testing Set Performance
y_pred
```

```python
from sklearn.metrics import classification_report, confusion_matrix

# we have used .2f for only 2 decimal places
print(f'Accuracy {100 * accuracy_score(y_pred, y_test):.2f} %')
```

```python
cm = confusion_matrix(y_pred, y_test)
sns.heatmap(cm, annot=True)
```

```python
print(classification_report(y_test, y_pred))
```

# TASK #6: TRAIN AND EVALUATE A RANDOM FOREST CLASSIFIER

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
# Testing Set Performance
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True)
```

```python
print(classification_report(y_test, y_pred))
```

# TASK #7: TRAIN AND EVALUATE A DEEP LEARNING MODEL

```python
import tensorflow as tf

model = tf.keras.models.Sequential()
model.add(tf.keras.layers.Input(shape=(50,)))
model.add(tf.keras.layers.Dense(units=500, activation='relu'))
model.add(tf.keras.layers.Dense(units=500, activation='relu'))
model.add(tf.keras.layers.Dense(units=500, activation='relu'))
model.add(tf.keras.layers.Dense(units=1, activation='sigmoid'))
```

```python
model.summary()
```

```python
model.compile(optimizer='Adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
epochs_hist = model.fit(X_train, y_train, epochs=100, batch_size=50)
```

```python
# oversampler = SMOTE(random_state=0)
# smote_train, smote_target = oversampler.fit_sample(X_train, y_train)
# epochs_hist = model.fit(smote_train, smote_target, epochs = 100, batch_size = 50)

y_predictions = model.predict(X_test)
y_predictions = (y_predictions > 0.5)
```

```python
y_predictions
```

```python
plt.plot(epochs_hist.history['loss'])
plt.title('Model Loss Progress During Training')
plt.xlabel('Epochs')
plt.ylabel('Training Loss')
plt.legend(['Training Loss'])
```

```python
plt.plot(epochs_hist.history['accuracy'])
plt.title('Model Accuracy Progress During Training')
plt.xlabel('Epoch')
plt.ylabel('Training Accuracy')
plt.legend(['Training Accuracy'])
```

```python
cm = confusion_matrix(y_test, y_predictions)
sns.heatmap(cm, annot=True)
```

```python
print(classification_report(y_test, y_predictions))
```
